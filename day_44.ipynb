```python
# Preserve original dataset
df_original = df.copy()

# Create binary classification target
df_original["target"] = (df_original["num"] > 0).astype(int)


# EXPERIMENT A: Remove rows with missing values
df_deleted = df_original.dropna(
    subset=["ca", "thal"]
).copy()


# EXPERIMENT B: Replace missing values
df_imputed = df_original.copy()

# Calculate replacement values
ca_mode = df_original["ca"].mode()[0]
thal_mode = df_original["thal"].mode()[0]

# Apply imputation
df_imputed["ca"] = df_imputed["ca"].fillna(ca_mode)
df_imputed["thal"] = df_imputed["thal"].fillna(thal_mode)
```

```python
# Experiment A
X_deleted = df_deleted.drop(columns=["num", "target"])
y_deleted = df_deleted["target"]


# Experiment B
X_imputed = df_imputed.drop(columns=["num", "target"])
y_imputed = df_imputed["target"]
```

```python
print("EXPERIMENT A")
print("Features:", X_deleted.shape)
print("Labels:", y_deleted.shape)
print("Missing values:", X_deleted.isnull().sum().sum())

print("\nEXPERIMENT B")
print("Features:", X_imputed.shape)
print("Labels:", y_imputed.shape)
print("Missing values:", X_imputed.isnull().sum().sum())
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components = 5)
X_pca = pca.fit_transform(xdel_scaled)
print(pca.explained_variance_ratio_)
```

```python
import numpy as np

cumulative_variance = np.cumsum(
    pca.explained_variance_ratio_
)

print(cumulative_variance)
```

```python
thresholds = [0.90, 0.95, 0.99]

for threshold in thresholds:
    components = np.argmax(
        cumulative_variance >= threshold
    ) + 1

    print(
        f"{threshold * 100}% variance: "
        f"{components} components"
    )
```

```python
import matplotlib.pyplot as plt

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o"
)

plt.axhline(y=0.90, linestyle="--", label="90%")
plt.axhline(y=0.95, linestyle="--", label="95%")
plt.axhline(y=0.99, linestyle="--", label="99%")

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA - Cumulative Explained Variance")

plt.xticks(range(1, 6))
plt.legend()
plt.grid()
plt.show()
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

scores = cross_val_score(
    rf,
    xdel_train,
    ydel_train,
    cv=skf,
    scoring="accuracy"
)

print("Fold scores:", scores)
print("Mean accuracy:", scores.mean())
print("Standard deviation:", scores.std())
```

```python
from sklearn.model_selection import cross_validate

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1",
    "roc_auc": "roc_auc"
}

results = cross_validate(
    rf,
    xdel_train,
    ydel_train,
    cv=skf,
    scoring=scoring,
    return_train_score=True
)


for metric in scoring:
    train_score = results[f"train_{metric}"].mean()
    val_score = results[f"test_{metric}"].mean()

    print(f"{metric}:")
    print(f"  Training: {train_score:.4f}")
    print(f"  Validation: {val_score:.4f}")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

numeric_features = [
    "age", "trestbps", "chol",
    "thalach", "oldpeak", "ca"
]

categorical_features = [
    "sex", "cp", "fbs",
    "restecg", "exang", "slope", "thal"
]

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ),

    "SVM": SVC(
        kernel="rbf"
    )
}
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

model_pipelines = {
    "Random Forest": Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(
            n_estimators=100,
            random_state=42
        ))
    ]),

    "SVM": Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", SVC(
            kernel="rbf"
        ))
    ])
}
```

```python
comparison = []

for name, pipeline in model_pipelines.items():

    results = cross_validate(
        pipeline,
        xdel_train,
        ydel_train,
        cv=skf,
        scoring=scoring,
        return_train_score=True
    )

    row = {"Model": name}

    for metric in scoring:
        row[metric] = results[f"test_{metric}"].mean()
        row[f"{metric}_std"] = results[f"test_{metric}"].std()

    comparison.append(row)

comparison_df = pd.DataFrame(comparison).set_index("Model")

print(comparison_df.round(4))
```

```python
from sklearn.model_selection import GridSearchCV
param_grids = {
    "Random Forest": {
        "classifier__n_estimators": [100, 200],
        "classifier__max_depth": [None, 3, 5, 8],
        "classifier__min_samples_split": [2, 5],
        "classifier__min_samples_leaf": [1, 2, 4]
    },

    "SVM": {
        "classifier__C": [0.1, 1, 10, 100],
        "classifier__gamma": ["scale", "auto", 0.01, 0.1]
    }
}

searches = {}

for name, pipeline in model_pipelines.items():

    grid = GridSearchCV(
        estimator=pipeline,
        param_grid=param_grids[name],
        scoring=scoring,
        refit="roc_auc",
        cv=skf,
        n_jobs=-1,
        return_train_score=True
    )

    grid.fit(xdel_train, ydel_train)

    searches[name] = grid

    print(f"\n{name}")
    print("Best parameters:", grid.best_params_)
    print("Best ROC-AUC:", grid.best_score_)

for name, grid in searches.items():

    print(f"\n{name}")
    best_idx = grid.best_index_

    for metric in scoring:
        mean = grid.cv_results_[f"mean_test_{metric}"][best_idx]
        std = grid.cv_results_[f"std_test_{metric}"][best_idx]

        print(f"{metric}: {mean:.4f} ± {std:.4f}")
```

```python
best_rf = searches["Random Forest"].best_estimator_


from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score
)

y_pred = best_rf.predict(xdel_test)
y_scores = best_rf.predict_proba(xdel_test)[:, 1]

print("Confusion Matrix:")
print(confusion_matrix(ydel_test, y_pred))

print("\nClassification Report:")
print(classification_report(ydel_test, y_pred))

print("Test ROC-AUC:", roc_auc_score(ydel_test, y_scores))
```

```python
from sklearn.model_selection import cross_val_predict

oof_probabilities = cross_val_predict(
    best_rf,
    xdel_train,
    ydel_train,
    cv=skf,
    method="predict_proba",
    n_jobs=-1
)[:, 1]

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

thresholds = [0.50, 0.45, 0.40, 0.35, 0.30]

threshold_results = []

for threshold in thresholds:

    predictions = (
        oof_probabilities >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        ydel_train,
        predictions,
        labels=[0, 1]
    ).ravel()

    specificity = tn / (tn + fp)

    threshold_results.append({
        "threshold": threshold,
        "accuracy": accuracy_score(ydel_train, predictions),
        "precision": precision_score(ydel_train, predictions, zero_division=0),
        "recall": recall_score(ydel_train, predictions),
        "f1": f1_score(ydel_train, predictions),
        "specificity": specificity,
        "false_negatives": fn,
        "false_positives": fp
    })

threshold_df = pd.DataFrame(threshold_results)

print(threshold_df.round(4))
```
